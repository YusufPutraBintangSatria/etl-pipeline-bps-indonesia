# EXTRACT DATA

```python
import pandas as pd
import sqlite3
print("Memulai proses Extract...")
# Pastikan nama file CSV-nya sama
df = pd.read_csv('data_bps_2021.csv')
```

# TRANSFORM DATA

```python
print("Membersihkan dan merapikan data...")

# Mengubah nama kolom bahasa Inggris jadi bahasa Indonesia & format standar database (huruf kecil pake underscore)
df.rename(columns={
    'province': 'provinsi',
    'cities_reg': 'kabupaten_kota',
    'poorpeople_percentage': 'persentase_miskin',
    'reg_gdp': 'pdrb', # Perputaran Ekonomi
    'life_exp': 'harapan_hidup',
    'avg_schooltime': 'rata_rata_sekolah',
    'exp_percap': 'pengeluaran_per_kapita'
}, inplace=True)

# Ubah ke string -> Hapus titik desimal -> Kembalikan jadi angka float
df['pdrb'] = df['pdrb'].astype(str).str.replace('.', '', regex=False).astype(float)

# Kalau ada data yang kosong (NaN), kita isi dengan angka 0 biar database nggak error
df.fillna(0, inplace=True)

print("\nPreview PDRB yang udah dibenerin (Jakarta harusnya udah puluhan ribu):")
# Kita intip data Jakarta Pusat buat bukti (Index 157)
print(df[df['kabupaten_kota'].str.contains("Jakarta Pusat")][['kabupaten_kota', 'pdrb']])

# Ngecek 5 baris pertama data yang udah bersih
print("\nPreview Data Bersih:")
print(df.head())
```

# LOAD DATA

```python
print("\nMenyimpan ke Database...")

# Bikin file database baru (atau numpuk yang lama kalau udah ada)
conn = sqlite3.connect('database_bps.db')

# Masukin dataframe df ke tabel bernama 'tabel_ekonomi'
df.to_sql('tabel_ekonomi', conn, if_exists='replace', index=False)

conn.close()
print("🔥 BERES BRO! Data BPS asli udah sukses masuk ke 'database_bps.db'!")
```
